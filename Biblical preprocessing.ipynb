```python
def cleanup(path):
    allchars = set()
    file = open(path)
    file = file.read()
    for word in file:
        allchars.update(char.lower() for char in word) #посмотрим символы до очистки: оставляем знаки препинания, чтобы поправить паузы вручную
    print(sorted(allchars), sep='')
    file = file.lower().replace('\n', ' ') #склеим текст
    file = re.sub('\s+', ' ', file).strip()
    file = file.replace('ÿ', 'ӈ') #правим кодировку
    file = file.replace('џ', 'ӈ')
    file = file.replace('¸', 'ё')
    file = file.replace('¨', 'ё')
    file = file.replace('& ', '') #убираем переносы
    file = file.replace('&', '-')
    start, end = file.find('1й пеляда'), file.find('вади” хамэди”') #откидываем вступление и словарь
    file = file[start:end]
    versepat = r'(\d+й\sпеляда)'
    file = re.sub(versepat, r'\n___\1___\n', file)
    file = re.sub(r'\*+\s\d+:\d+[а-я\s’,«»”.ӈ]+', '', file) #убираем сноски
    file = re.sub(r'(марк|иоанн)\s\d+(,\s\d+)?', '', file) #убираем названия страниц в начале
    splitpat = r'([.!?:])\s*' #делим по строкам так, где скорее всего будут паузы
    file = re.sub(splitpat, r'\1\n', file)
    file = re.sub('\d+\s', '', file) #чистим от чисел
    newchars = set() #посмотрим на символы после очистки:
    for word in file:
        newchars.update(char.lower() for char in word)
    print(sorted(newchars), sep='')
    newname = path.replace('.txt', '_new.txt')
    with open(newname, 'w') as ffile:
        ffile.write(file)
    print(f'Создан файл {newname}')
```

```python
def split_text_into_verses(txt_path):
    with open(txt_path, 'r', encoding='utf-8') as f:
        text = f.read()
    verses = {}
    current_verse = None
    for line in text.split('\n'):
        line = line.strip()
        if line.startswith('___') and line.endswith('___'): #ищем начало
            verse_num = line[3:-3].split('й пеляда')[0].strip().replace('‑', '')
            verse_num = '0' + verse_num if len(verse_num) == 2 else '00' + verse_num
            current_verse = verse_num
            verses[current_verse] = []
        elif current_verse is not None and line:
            verses[current_verse].append(line)
    return verses
```

```python
def update_textgrid_with_text(textgrid_path, lines):
    tg = tgt.read_textgrid(textgrid_path, encoding="utf-8")
    tier = tg.get_tier_by_name('текст')
    if len(lines) != len([i for i in tier if i.text.strip() != ""]):
        print(f'Количество непустых интервалов в {textgrid_path} не совпадает с количеством строк: \n {len(lines)} строк, {len([i for i in tier if i.text.strip() != ""])} интервалов')
        return
    for i, interval in enumerate(tier):  #проходим по интервалам и заменяем текст на строки из txt
        if i < len(lines): 
            if interval.text.strip() == "":
                continue                #пропускаем пустые интервалы
            new_text = lines[i]         #заменяем текст интервала на соответствующую строку
            tier[i].text = new_text

    tgt.io.write_to_file(tg, textgrid_path, format="long")
    print(f"Обновлён файл: {textgrid_path}")
```

```python
def process_all_verses(txt_path, textgrid_dir):
    verses = split_text_into_verses(txt_path)
    for verse_num, lines in verses.items():
        textgrid_path = os.path.join(textgrid_dir, f"{verse_num}.TextGrid")
        if os.path.exists(textgrid_path):
            update_textgrid_with_text(textgrid_path, lines)
        else:
            print(f"Файл не найден: {textgrid_path}")
```

```python
def pd_from_tgt(path):
    data = []
    for file in os.listdir(path):
        if file.endswith('.TextGrid'):
            tg = tgt.io.read_textgrid(os.path.join(path, file))
            tier = tg.get_tier_by_name('текст')
            for i, interval in enumerate(tier):
                    data.append({
                        "file_name" : os.path.join(path.replace('tgt', 'wav'), file.replace('TextGrid', 'wav')),
                        "start_time": interval.start_time,
                        "end_time": interval.end_time,
                        "label": interval.text,
                        "length": interval.end_time - interval.start_time
                    })
    df = pd.DataFrame(data)
    df.to_csv(path.replace(' tgt', '.csv'), index=False)
    return df
```

```python
def calculate_time(dataframe):
    print(f'Средняя длина интервала {dataframe['length'].mean()}, максимальная {dataframe['length'].max()}, минимальная {dataframe['length'].min()}')
    print(f'Звучащее время {dataframe['length'].sum() // 3600}ч {dataframe['length'].sum() % 3600 // 60}мин {dataframe['length'].sum() % 60}с')
```

```python
mark_df = pd_from_tgt('/Users/air/Downloads/Ненецкие аудио/Библейские ненецкие аудио/Mark/Mark old tgt')
calculate_time(mark_df)
```
