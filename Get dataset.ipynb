```python
import pydub
import pandas as pd
from pydub import AudioSegment
import os
```

```python
def slice_audio(row, output_dir):
    try:
        audio = AudioSegment.from_file(row['file_name'])
        start, end = row['start_time'] * 1000, row['end_time'] * 1000
        sliced = audio[start:end]
        base_name = os.path.basename(row['file_name']).split('.')[0]
        new_file_name = f"{base_name}_{start}_{end}.wav"
        new_file_path = os.path.join(output_dir, new_file_name)
        sliced.export(new_file_path, format="wav")
        return new_file_path
    except Exception as e:
        print(f"Ошибка при обработке {row['file_name']}: {e}")
        return None
```

```python
mark = pd.read_csv('/Users/air/Documents/PycharmProjects/Nenets/mark.csv')
john = pd.read_csv('/Users/air/Documents/PycharmProjects/Nenets/john.csv')
jonah = pd.read_csv('/Users/air/Documents/PycharmProjects/Nenets/jonah.csv')
parables = pd.read_csv('/Users/air/Documents/PycharmProjects/Nenets/parables.csv')
chumoteka = pd.read_csv('/Users/air/Downloads/Ненецкие аудио/Чумотека/Чумотека.csv')
siberian = pd.read_csv('/Users/air/Documents/PycharmProjects/Nenets/siberian.csv')
asr_dataset = pd.concat([mark, john, jonah, parables, chumoteka, siberian], ignore_index=True)
```

```python
#выкинем слишком короткие и слишком длинные отрезки
asr_dataset = asr_dataset[asr_dataset['length'] < 11]
asr_dataset = asr_dataset[asr_dataset['length'] > 3]
asr_dataset[:10]
```

```python
unique_chars = sorted(list(set(asr_dataset['label'].str.cat()))) #подчистим табличку
print(unique_chars)
```

```python
replace_table = {'ң':'ӈ', '’':"'", '“':'"', '”':'"', '′':"'", '‹':'', '›':'', '!':'', '(':'', ')':'', '*':'', ',':'',
                  '-':'', '.':'', '0':'', '1':'', '4':'', '5':'', '6':'', ':':'', ';':'', '?':'', '[':'', ']':'', 'a':'а',
                   '«':'', '»':'', '̈':'', '‑':' ', '–':'', '  ':' ', '\n':'', '   ':' '}
asr_dataset['label'] = asr_dataset['label'].str.replace(replace_table)
asr_dataset['label'] = asr_dataset['label'].str.lower()
```

```python
after_cleanup = sorted(list(set(asr_dataset['label'].str.cat()))) #остались только буквы, тасеры и пробел
print(after_cleanup)
```

```python
output_dir = '/Users/air/Downloads/Ненецкие аудио/data'
asr_dataset['sliced_file_path'] = asr_dataset.apply(lambda row: slice_audio(row, output_dir), axis=1)
sliced_dataset = asr_dataset[['sliced_file_path', 'label']].copy()
sliced_dataset.columns = ['audio', 'transcription']
sliced_dataset.to_csv("metadata.csv", index=False)
```
