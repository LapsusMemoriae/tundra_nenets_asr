```python
import os
import wave
from pydub import AudioSegment
from tinytag import TinyTag
```

```python
exts = ('.mp3','.ogg', '.wav')

for root, dirs, files in os.walk('/Users/air/Downloads/Ненецкие аудио/'):
    for file in files:
        if file.lower().endswith(exts):
            file_path = os.path.join(root, file)
            try:
                sound = AudioSegment.from_file(file_path)
                sound = sound.set_channels(1)
                sound = sound.set_frame_rate(16000)
                sound = sound.set_sample_width(2)
                output_path = os.path.splitext(file_path)[0] + '.wav'
                sound.export(output_path, format="wav")
                print(f"Converted: {file_path} -> {output_path}")
            except Exception as e:
                print(f"Failed to process {file_path}: {e}")
```

```python
use = ['/Users/air/Downloads/Ненецкие аудио/Чумотека', '/Users/air/Downloads/Ненецкие аудио/Библейские ненецкие аудио', '/Users/air/Downloads/Ненецкие аудио/Siberian languages']
#проверка длительности аудио
total_seconds = 0
for i in use:
    for root, dirs, files in os.walk(i):
        for file in files:
            if file.endswith(('.wav')):
                tag = TinyTag.get(os.path.join(root, file))
                total_seconds += tag.duration
            #print(dir, total_seconds)

print(f"Всего: {total_seconds // 3600} часов {total_seconds % 3600 // 60} минут {total_seconds % 60} секунд")
```
